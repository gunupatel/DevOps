# Monitoring

Monitoring is the process of continuously observing infrastructure, applications, and services to track their health, performance, and availability. In DevOps, tools like Prometheus, Grafana, and Alertmanager are commonly used to collect metrics, visualize data, and generate alerts for proactive issue resolution.

Without Monitoring :
Application Down
      ↓
Customers Complain
      ↓
Team Investigates

With Monitoring :
Application Down
      ↓
Alert Generated
      ↓
DevOps Team Notified
      ↓
Issue Fixed Quickly

# Popular Monitoring Tools
1. Prometheus : Prometheus is an open-source monitoring and alerting tool used to collect and store metrics from applications and infrastructure.

Used for:
Collecting metrics
Storing time-series data
Alerting

Examples:
CPU Metrics
Memory Metrics
Pod Metrics

2. Grafana : Grafana is a visualization tool used to create dashboards and display metrics collected by Prometheus.

Used for:
Dashboards
Visualization
Monitoring Reports

Example Dashboard:
CPU Usage
Memory Usage
Node Status
Pod Status

3. Alertmanager : Alertmanager handles alerts generated by Prometheus and sends notifications through email, Slack, Teams, etc.

Used for:
Sending alerts
Email notifications
Slack notifications


4. Node Exporter : Node Exporter collects server-level metrics such as CPU, memory, disk, and network information and exposes them to Prometheus.

Collects server metrics such as:
CPU
Memory
Disk
Network

# Logging

Logging is the process of collecting, storing, and analyzing application and infrastructure logs to troubleshoot issues, perform root cause analysis, and gain visibility into system behavior. Common tools include ELK Stack, EFK Stack, Loki, and Splunk.

Without logs:
Application crashes
       ↓
No information available
       ↓
Difficult troubleshooting

With logs: 
Application crashes
       ↓
Check logs
       ↓
Identify root cause
       ↓
Fix issue quickly

Types of Logs
1. Application Logs : Generated by applications.

Example:
Login Success
Payment Failed
Database Connectivity Error

2. System Logs : Generated by Linux OS.

Location:
/var/log/messages

3. Web Server Logs
Nginx :
/var/log/nginx/access.log

4. Kubernetes Logs

Check Pods:
kubectl get pods

View pod logs:
kubectl logs pod-name

View previous logs:
kubectl logs nginx-pod --previous

Common Commands for Logs
1. View Entire File : cat app.log
2. Last 10 Lines : tail app.log
3. Monitor Logs in Real Time : tail -f app.log
4. Search for Errors : grep ERROR app.log
5. Count Errors : grep ERROR app.log | wc -l

# Logging Tools in DevOps
ELK Stack

Elasticsearch : Stores logs.

Logstash : Collects and processes logs.

Kibana : Visualizes logs.


NOTE : Logging is used to record information about the events that occur within a system or application, such as errors and user actions. This information can be used for troubleshooting, auditing, and compliance purposes.
On the other hand, Monitoring is used to track the performance and health of a system or application, such as resource usage, network traffic, and error rates. This information can be used to identify and diagnose issues and to predict and prevent problems before they occur.
